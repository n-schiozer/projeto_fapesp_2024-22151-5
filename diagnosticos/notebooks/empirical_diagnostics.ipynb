Este notebook não baixa nem transforma dados. Execute primeiro o pipeline em empirical/src/build_empirical_base.py; aqui são lidas apenas as interfaces finais.

```python
from pathlib import Path
import pandas as pd

EMPIRICAL = Path.cwd()
if not (EMPIRICAL / 'data').exists():
    EMPIRICAL = Path.cwd() / 'empirical'

targets = pd.read_csv(EMPIRICAL / 'data' / 'processed' / 'model_targets.csv')
moments = pd.read_csv(EMPIRICAL / 'outputs' / 'empirical_moments.csv')
catalog = pd.read_csv(EMPIRICAL / 'metadata' / 'series_catalog.csv')
```

```python
catalog.groupby(['block', 'status']).size().rename('n_series').reset_index()
```

```python
moments.pivot_table(index=['series_id', 'model_variable'], columns='moment', values='value')
```

```python
targets.assign(date=pd.to_datetime(targets['date'])).groupby('series_id').agg(
    first_date=('date', 'min'), last_date=('date', 'max'), n_obs=('value', 'size')
)
```
